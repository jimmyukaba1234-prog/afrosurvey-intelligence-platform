.config("spark.hadoop.fs.s3a.access.key", config["storage"]["minio"]["access_key"])
.config("spark.hadoop.fs.s3a.secret.key", config["storage"]["minio"]["secret_key"])
.config("spark.hadoop.fs.s3a.path.style.access", "true")
.config("spark.hadoop.fs.s3a.impl", "org.apache.hadoop.fs.s3a.S3AFileSystem")
.config("spark.hadoop.fs.s3a.connection.ssl.enabled", "false")
.config("spark.hadoop.mapreduce.fileoutputcommitter.algorithm.version", "2")
Why Hadoop appears in the project

Even though the project is using Spark, Spark still relies on Hadoop libraries for filesystem access.

MinIO behaves like S3.

Spark cannot talk to S3/MinIO directly by itself. It uses Hadoop’s S3A connector.

So the flow is:

Spark
  ↓
Hadoop S3A connector
  ↓
MinIO

That is why packages like these were needed:

org.apache.hadoop:hadoop-aws:3.3.4
com.amazonaws:aws-java-sdk-bundle:1.12.262
Simple explanation of Hadoop

Hadoop is like a big system that lets many computers store and process huge data together.

In this project, we are not using full Hadoop HDFS, but we are using parts of the Hadoop ecosystem.

Specifically, we use Hadoop’s S3A connector so Spark can read and write data from MinIO.
